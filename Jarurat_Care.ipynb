```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

base_url = "http://books.toscrape.com/catalogue/page-{}.html"

data = []

for page in range(1, 11):  # 10 pages = ~200 entries
    url = base_url.format(page)
    print(f"Scraping page {page}...")

    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    books = soup.find_all("article", class_="product_pod")

    print("Books found:", len(books))

    for book in books:
        name = book.h3.a["title"]

        data.append({
            "Name": name,
            "Email": "Not Available",
            "Website": "http://books.toscrape.com",
            "Location": "UK"
        })

df = pd.DataFrame(data)

print("\nTotal rows scraped:", df.shape[0])
df.head()
```

```python
df.drop_duplicates(inplace=True)
df.fillna("Not Available", inplace=True)

def generate_email(row):
    name = str(row["Name"]).lower().replace(" ", ".")
    return f"{name}@bookstore.com"

df["Generated Email"] = df.apply(generate_email, axis=1)
```

```python
df.to_excel("leads.xlsx", index=False)

from google.colab import files
files.download("leads.xlsx")
```
